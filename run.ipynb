```python
from jetbot import Robot
import time

robot = Robot()

TURN_SPEED = 0.4
DRIVE_SPEED = 0.1
SECONDS_PER_DEGREE = 0.0012

import time

def drive_forward(robot, duration, max_speed=DRIVE_SPEED, ramp_time=0.2, tick=0.01):
    steps = int(ramp_time / tick)
    speed_increment = max_speed / steps

    # --- Ramp up ---
    speed = 0
    for _ in range(steps):
        speed += speed_increment
        if speed > max_speed:
            speed = max_speed
        robot.set_motors(speed, speed)
        time.sleep(tick)

    # --- Constant speed ---
    cruise_time = duration - 2 * ramp_time
    if cruise_time > 0:
        robot.set_motors(max_speed, max_speed)
        time.sleep(cruise_time)

    # --- Ramp down ---
    speed = max_speed
    for _ in range(steps):
        speed -= speed_increment
        if speed < 0:
            speed = 0
        robot.set_motors(speed, speed)
        time.sleep(tick)

    robot.stop()


def drive_backward(robot, duration):
    robot.set_motors(-DRIVE_SPEED, -DRIVE_SPEED)
    time.sleep(duration)
    robot.stop()

def turn_left(robot, degrees):
    turn_time = degrees * SECONDS_PER_DEGREE
    robot.set_motors(-TURN_SPEED, TURN_SPEED)
    time.sleep(turn_time)
    robot.stop()

def turn_right(robot, degrees):
    turn_time = degrees * SECONDS_PER_DEGREE
    robot.set_motors(TURN_SPEED, -TURN_SPEED)
    time.sleep(turn_time)
    robot.stop()
```
